# Dataset Overview — Telco Customer Churn

## Columns — What Each One Is and How We Used It

| Column | What It Means | Type | Did We Use It? |
|---|---|---|---|
| `customerID` | Unique ID for each customer | Text | No — just an identifier, not useful for analysis |
| `gender` | Male / Female | Categorical | No — not used in our models |
| `SeniorCitizen` | Is the customer 65+? (0/1) | Categorical | Yes — used in Naive Bayes & Decision Tree |
| `Partner` | Has a spouse/partner? (Yes/No) | Categorical | No |
| `Dependents` | Has dependents living with them? (Yes/No) | Categorical | No |
| `tenure` | Number of months as a customer | Numeric | **Yes — one of our strongest features** (correlation -0.35 with churn) |
| `PhoneService` | Has phone service? (Yes/No) | Categorical | No |
| `MultipleLines` | Has multiple phone lines? | Categorical | No |
| `InternetService` | DSL / Fiber optic / No internet | Categorical | **Yes — used in EDA and both models** |
| `OnlineSecurity` | Has online security add-on? | Categorical | No |
| `OnlineBackup` | Has online backup add-on? | Categorical | No |
| `DeviceProtection` | Has device protection add-on? | Categorical | No |
| `TechSupport` | Has tech support add-on? | Categorical | No |
| `StreamingTV` | Has streaming TV add-on? | Categorical | No |
| `StreamingMovies` | Has streaming movies add-on? | Categorical | No |
| `Contract` | Month-to-month / One year / Two year | Categorical | **Yes — our single strongest feature** (58.8% model importance) |
| `PaperlessBilling` | Uses paperless billing? (Yes/No) | Categorical | No |
| `PaymentMethod` | How they pay (e-check, credit card, etc.) | Categorical | Yes — used in models, but turned out to have ~0 importance |
| `MonthlyCharges` | What they pay per month ($) | Numeric | **Yes — used in EDA, stats, and Decision Tree** |
| `TotalCharges` | Total amount billed so far ($) | Numeric | Yes — used in EDA/stats; needed cleaning (was loaded as text) |
| `Churn` | Did they leave? (Yes/No) | Categorical | **This is our target — what we're predicting** |

## Quick Grouping

| Group | Columns |
|---|---|
| **Target (what we predict)** | `Churn` |
| **Core features we actually used** | `Contract`, `tenure`, `MonthlyCharges`, `InternetService`, `SeniorCitizen`, `PaymentMethod` |
| **Identifier (not analyzed)** | `customerID` |
| **Unused demographic/service columns** | `gender`, `Partner`, `Dependents`, `PhoneService`, `MultipleLines`, `OnlineSecurity`, `OnlineBackup`, `DeviceProtection`, `TechSupport`, `StreamingTV`, `StreamingMovies`, `PaperlessBilling` |
| **Money column needing cleanup** | `TotalCharges` (was text, converted to numeric) |

**Why we didn't use every column:** We picked the features that EDA showed had the strongest link to churn, to keep the model focused and explainable — a common real-world practice, since using every column doesn't always improve a model and can add noise.
